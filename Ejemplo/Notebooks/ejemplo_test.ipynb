#### primero se importan los módulos a probar. Ya que queremos comprobar si el código que hicimos funciona correctamente

```python
from Ejemplo.Modulos.extraction.pdf_extractor import extract_text_from_pdf
from Ejemplo.Modulos.nlp.text_classifier import split_qa
```

#### 1. Extraer texto

```python
texto = extract_text_from_pdf("data/Habilidades_Vida/ejemplo.pdf")
print(texto)
```

#### 2. Separar preguntas y respuestas

```python
qa = split_qa(texto)
print("Preguntas encontradas:", qa)
```

#### 3. Assert simple

```python
# el assert es para verificar que se extrajo al menos una pregunta y respuesta
# y que no están vacías
assert len(qa) > 0, "No se extrajo ninguna pregunta."

# esto es para mostrar la pregunta y con su respuesta
print("Preguntas y respuestas extraídas:")
for i, (question, answer) in enumerate(qa):
    print(f"Pregunta {i + 1}: {question}")
    print(f"Respuesta {i + 1}: {answer}")
    print()
    assert question != "", "La pregunta está vacía."
    assert answer != "", "La respuesta está vacía."
```
